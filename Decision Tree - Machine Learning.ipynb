## Diskritisasi Fitur
Ambang batas identik dengan tabel di bagian atas sheet Excel.

```python
# Ambang (Batas1, Batas2): < B1 = Rendah ; B1..B2 = Sedang ; > B2 = Tinggi
AMBANG = {
    'Sangat Pendek': (517, 904),
    'Pendek':        (1564, 2692),
    'Normal':        (12009, 19000),
    'Tinggi':        (16, 38),
}

def diskritisasi(nilai, batas):
    b1, b2 = batas
    if nilai < b1: return 'Rendah'
    elif nilai <= b2: return 'Sedang'
    else: return 'Tinggi'

for fitur, batas in AMBANG.items():
    df[fitur + '_kat'] = df[fitur].apply(lambda v: diskritisasi(v, batas))

df[['Wilayah','Sangat Pendek_kat','Pendek_kat','Normal_kat','Tinggi_kat','Kelas']].head(10)
```

## B. Entropy Data Awal
$$Entropy(S) = -\sum p_i \cdot \log_2(p_i)$$

```python
KELAS = ['Aman','Waspada','Kritis']

def entropy(jumlah_per_kelas):
    total = sum(jumlah_per_kelas)
    if total == 0: return 0.0
    e = 0.0
    for c in jumlah_per_kelas:
        if c > 0:
            p = c/total
            e -= p*math.log2(p)
    return e

jml_total = [(df['Kelas']==c).sum() for c in KELAS]
N = len(df)
entropy_S = entropy(jml_total)

print('Jumlah per kelas :', dict(zip(KELAS, jml_total)))
for c, j in zip(KELAS, jml_total):
    print(f'  p({c}) = {j}/{N} = {j/N:.4f}')
print(f'\nEntropy(S) = {entropy_S:.4f}')
```

## C. Information Gain Tiap Fitur
$$Gain(A) = Entropy(S) - \sum \frac{|S_v|}{|S|}\,Entropy(S_v)$$

```python
FITUR = ['Sangat Pendek','Pendek','Normal','Tinggi']
NILAI = ['Rendah','Sedang','Tinggi']
gain = {}

for fitur in FITUR:
    kol = fitur + '_kat'
    print(f'=== Fitur: {fitur} ===')
    entropy_setelah = 0.0
    for v in NILAI:
        cc = [((df[kol]==v) & (df['Kelas']==c)).sum() for c in KELAS]
        sv = sum(cc)
        ent_v = entropy(cc)
        bobot = sv/N
        entropy_setelah += bobot*ent_v
        print(f'  {v:7s} Aman={cc[0]:2d} Waspada={cc[1]:2d} Kritis={cc[2]:2d} '
              f'|Sv|={sv:2d} Entropy(Sv)={ent_v:.4f} bobot*ent={bobot*ent_v:.4f}')
    gain[fitur] = entropy_S - entropy_setelah
    print(f'  Entropy setelah split = {entropy_setelah:.4f}')
    print(f'  GAIN = {entropy_S:.4f} - {entropy_setelah:.4f} = {gain[fitur]:.4f}\n')
```

## D. Pemilihan Akar (Gain tertinggi)

```python
print('Information Gain tiap fitur:')
for f in FITUR:
    print(f'  {f:15s} : {gain[f]:.4f}')

akar = max(gain, key=gain.get)
print(f'\nGain tertinggi   : {gain[akar]:.4f}')
print(f'AKAR POHON       : {akar}')
```
